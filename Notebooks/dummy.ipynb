```python
from sklearn.preprocessing import StandardScaler
import pandas as pd
import joblib

# Load dataset
df = pd.read_csv("/Users/siddharthaganguli/Desktop/Laptop-Price-Pred/data/laptops.csv")

# Create scalers
feature_scaler = StandardScaler()
price_scaler = StandardScaler()

# Scale only numeric features (excluding Final Price)
numeric_features = ['RAM', 'Storage', 'Screen']
df[numeric_features] = feature_scaler.fit_transform(df[numeric_features])

# Scale target separately
df['Final Price'] = price_scaler.fit_transform(df[['Final Price']])

# Save both scalers
joblib.dump(feature_scaler, "models/feature_scaler.pkl")
joblib.dump(price_scaler, "models/price_scaler.pkl")

# Save dataset
df.to_csv("/Users/siddharthaganguli/Desktop/Laptop-Price-Pred/data/cleaned_scaled_laptops.csv", index=False)

print("✅ Feature and Price scalers saved successfully.")
```
